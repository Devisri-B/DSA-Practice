# List Operations 

````python
a = [3, 1, 4, 1, 5]
# Slicing
a[1:3]  # [1, 4] — from index 1 up to (not including) 3
a[::-1] # [5, 1, 4, 1, 3] — reversed
a[-1]  # 5 — last element

# Sorting
a.sort() # in-place
sorted(a, reverse=True) # returns new list, descending
sorted(a, key=lambda x: -x) # custom sort

# Useful methods
a.append(9) # add to end
a.insert(2, 99) # insert at index 2
a.pop() # remove and return last
a.pop(0) # remove and return first (slow for large lists)
a.count(1) # count occurrences of 1
a.index(4) # first index of value 4
sum(a) # sum of all
min(a), max(a) # min and max
````

# zip()

## What is zip()?

`zip()` combines multiple lists together, element by element, creating tuples. Think of it like a **zipper** — it weaves lists together.

```
List 1:  ['a', 'b', 'c']
List 2:  [1, 2, 3]
         ↓ ↓ ↓
zip()  → [('a', 1), ('b', 2), ('c', 3)]
```

## Basic Syntax

```python
zip(list1, list2, list3, ...)
```

Returns an **iterator of tuples** where each tuple contains one element from each list.

## Step-by-Step Example

```python
names = ['Alice', 'Bob', 'Charlie']
ages = [25, 30, 35]

result = list(zip(names, ages))
# [('Alice', 25), ('Bob', 30), ('Charlie', 35)]
```

## Most Common Use: Parallel Iteration

### WITHOUT zip (awkward)
```python
names = ['Alice', 'Bob']
ages = [25, 30]

for i in range(len(names)):
    print(f"{names[i]} is {ages[i]} years old")
```

### WITH zip (clean and Pythonic)
```python
names = ['Alice', 'Bob']
ages = [25, 30]

for name, age in zip(names, ages):
    print(f"{name} is {age} years old")
# Output:
# Alice is 25 years old
# Bob is 30 years old
```

## Three or More Lists

```python
names = ['Alice', 'Bob', 'Charlie']
ages = [25, 30, 35]
cities = ['NYC', 'LA', 'Chicago']

for name, age, city in zip(names, ages, cities):
    print(f"{name}, {age}, {city}")
# Output:
# Alice, 25, NYC
# Bob, 30, LA
# Charlie, 35, Chicago
```

## What if Lists Have Different Lengths?

`zip()` **stops at the shortest list**:

```python
list1 = ['a', 'b', 'c', 'd']
list2 = [1, 2, 3]

result = list(zip(list1, list2))
# [('a', 1), ('b', 2), ('c', 3)]
# Note: 'd' is ignored
```

**To keep all elements, use `zip_longest()`:**

```python
from itertools import zip_longest

list1 = ['a', 'b', 'c', 'd']
list2 = [1, 2, 3]

result = list(zip_longest(list1, list2, fillvalue=None))
# [('a', 1), ('b', 2), ('c', 3), ('d', None)]
```

## Practical Use Cases

### 1. Create a Dictionary from Two Lists

```python
keys = ['name', 'age', 'city']
values = ['Alice', 25, 'NYC']

data = dict(zip(keys, values))
# {'name': 'Alice', 'age': 25, 'city': 'NYC'}
```

### 2. Transpose a Matrix

```python
matrix = [[1, 2, 3], [4, 5, 6]]
transposed = list(zip(*matrix))
# [(1, 4), (2, 5), (3, 6)]
```

The `*` unpacks the matrix into separate arguments.

### 3. Pair Up Elements

```python
numbers = [1, 2, 3, 4, 5, 6]
pairs = list(zip(numbers[::2], numbers[1::2]))
# [(1, 2), (3, 4), (5, 6)]
```

### 4. Compare Two Lists

```python
list1 = [1, 2, 3]
list2 = [1, 2, 4]

for a, b in zip(list1, list2):
    if a != b:
        print(f"Difference: {a} vs {b}")
# Output: Difference: 3 vs 4
```

### 5. Create Variables from Lists

```python
data = list(zip([1, 2, 3], ['a', 'b', 'c'], [True, False, True]))
numbers, letters, bools = zip(*data)

print(numbers)  # (1, 2, 3)
print(letters)  # ('a', 'b', 'c')
print(bools)    # (True, False, True)
```

## Important Points

### 1. Returns an Iterator (Not a List)

```python
result = zip([1, 2], ['a', 'b'])
print(result)  # <zip object at 0x...>

# Can iterate once:
for item in result:
    print(item)

# But can't iterate again - exhausted!
# Convert to list if needed multiple times:
result = list(zip([1, 2], ['a', 'b']))
```

### 2. Returns Tuples (Not Lists)

```python
result = list(zip([1, 2], ['a', 'b']))
# [(1, 'a'), (2, 'b')] ← Each element is a tuple

# Each item is immutable:
first_pair = result[0]
print(type(first_pair))  # <class 'tuple'>
```

### 3. Works with Any Iterable

```python
# Works with strings, ranges, sets, etc.
list(zip("abc", range(1, 4)))
# [('a', 1), ('b', 2), ('c', 3)]

list(zip([1, 2], "xy"))
# [(1, 'x'), (2, 'y')]
```

## Common Gotchas

### Gotcha 1: Forgetting to Convert to List
```python
result = zip([1, 2, 3], ['a', 'b', 'c'])

# This doesn't work:
print(result[0])  # TypeError - not subscriptable

# Must convert first:
result = list(zip([1, 2, 3], ['a', 'b', 'c']))
print(result[0])  # (1, 'a')
```

### Gotcha 2: Iterator Exhaustion
```python
result = zip([1, 2, 3], ['a', 'b', 'c'])

list1 = list(result)  # Exhausts iterator
list2 = list(result)  # Empty!

# Solution: Create list first
result = list(zip([1, 2, 3], ['a', 'b', 'c']))
list1 = result
list2 = result
```

### Gotcha 3: Silently Truncating
```python
list1 = [1, 2, 3, 4, 5]
list2 = ['a', 'b']

result = list(zip(list1, list2))
# [(1, 'a'), (2, 'b')] ← Lost 3, 4, 5!
```

### Gotcha 4: Unpacking with * Can Be Confusing
```python
matrix = [[1, 2, 3], [4, 5, 6]]

# This:
list(zip(*matrix))

# Is equivalent to:
list(zip([1, 2, 3], [4, 5, 6]))

# The * unpacks each inner list as a separate argument
```

## Time and Space Complexity

| Operation | Complexity | Notes |
|-----------|-----------|-------|
| Creating zip object | O(1) | Just creates iterator |
| Converting to list | O(n) | n is shortest list length |
| Iterating | O(n) | Per element |
| Memory (iterator) | O(1) | Lazy evaluation |
| Memory (as list) | O(n) | If converted to list |

## Best Practices

1. Use `zip()` when iterating through multiple lists in parallel
2. Prefer `zip()` over index-based loops with `range(len())`
3. Use `enumerate()` if you need both index and value
4. Use `zip_longest()` if lists have different lengths
5. Remember `zip()` stops at shortest list by default
6. Convert to list only when you need it multiple times
7. Use for creating dictionaries from parallel lists
8. Use `*` unpacking for transposing matrices
9. Don't modify lists while zipping them
10. Remember each tuple is immutable

# deepcopy - Complete Guide

## What is deepcopy?

`deepcopy` (deep copy) creates a **completely independent copy** of an object, including all nested objects inside it. It goes "deep" into the structure and copies everything.

```python
from copy import deepcopy
```

This imports the `deepcopy` function from Python's `copy` module.

---

## Shallow Copy vs Deep Copy

### Shallow Copy (Regular Copy)

A **shallow copy** copies only the **outer layer**. Nested objects are still **shared** (aliased):

```python
original = [[1, 2], [3, 4]]
shallow = original.copy()  # or original[:]

# Modify nested element in copy
shallow[0][0] = 999

print(original)  # [[999, 2], [3, 4]] - CHANGED!
print(shallow)   # [[999, 2], [3, 4]]

# Both point to the same inner list!
```

**Visual:**
```
original: ┌─────────┐
          │  [0]  ──┼──→ [1, 2] ← shared reference
          │  [1]  ──┼──→ [3, 4] ← shared reference
          └─────────┘
          
shallow:  ┌─────────┐
          │  [0]  ──┼──→ [1, 2] ← SAME object
          │  [1]  ──┼──→ [3, 4] ← SAME object
          └─────────┘
```

### Deep Copy

A **deep copy** creates **completely separate copies** of everything, including nested objects:

```python
from copy import deepcopy

original = [[1, 2], [3, 4]]
deep = deepcopy(original)

# Modify nested element in copy
deep[0][0] = 999

print(original)  # [[1, 2], [3, 4]] - UNCHANGED!
print(deep)      # [[999, 2], [3, 4]]

# They are completely independent!
```

**Visual:**
```
original: ┌─────────┐
          │  [0]  ──┼──→ [1, 2] ← independent
          │  [1]  ──┼──→ [3, 4] ← independent
          └─────────┘
          
deep:     ┌─────────┐
          │  [0]  ──┼──→ [999, 2] ← separate object
          │  [1]  ──┼──→ [3, 4] ← separate object
          └─────────┘
```

---

## Real-World Example

```python
from copy import deepcopy

# Create a student profile with nested data
student = {
    'name': 'Alice',
    'scores': [85, 90, 88],
    'address': {'city': 'NYC', 'zip': '10001'}
}

# SHALLOW COPY - BAD for nested data
student_shallow = student.copy()
student_shallow['scores'][0] = 100

print(student['scores'])  # [100, 90, 88] - CHANGED!

# DEEP COPY - SAFE for nested data
student_deep = deepcopy(student)
student_deep['scores'][0] = 100

print(student['scores'])  # [85, 90, 88] - UNCHANGED!
```

---

## When Each Copy Is Used

### Use Shallow Copy When:
- Copying a **simple list** with only primitive values (numbers, strings)
- You don't have nested objects
- Speed is critical and data is simple

```python
# Fine with shallow copy
numbers = [1, 2, 3, 4, 5]
copy1 = numbers.copy()
copy1[0] = 999
print(numbers)  # [1, 2, 3, 4, 5] - safe because no nesting
```

### Use Deep Copy When:
- Copying lists/dicts with **nested structures**
- You need complete independence
- You don't want any shared references

```python
# MUST use deep copy
matrix = [[1, 2], [3, 4]]
copy1 = deepcopy(matrix)
copy1[0][0] = 999
print(matrix)  # [[1, 2], [3, 4]] - safe!
```

---

## Copy Methods Comparison

| Method | Type | Shared References? | When to Use |
|--------|------|-------------------|------------|
| `lst[:]` | Shallow | Yes (nested) | Simple lists |
| `lst.copy()` | Shallow | Yes (nested) | Simple lists |
| `list(lst)` | Shallow | Yes (nested) | Simple lists |
| `deepcopy(lst)` | Deep | NO | Nested structures |

## Step-by-Step: How deepcopy Works

```python
from copy import deepcopy

original = {
    'name': 'Alice',
    'grades': [90, 85, 88]
}

# Step 1: Create new outer dictionary
copy = {}

# Step 2: For each key-value pair, RECURSIVELY copy the value
#   - 'name': 'Alice' → copy as is (string is immutable)
#   - 'grades': [90, 85, 88] → RECURSIVELY copy the list
#     - [90, 85, 88] → [90, 85, 88] (new list)

# Result:
deep = deepcopy(original)
# Now 'deep' is completely independent!

deep['grades'][0] = 999
print(original['grades'])  # [90, 85, 88] - unchanged
print(deep['grades'])      # [999, 85, 88]
```

## Common Mistakes

### Mistake 1: Forgetting to use deepcopy with nested data

```python
# WRONG - Shallow copy with nested data
matrix = [[1, 2], [3, 4]]
copy = matrix[:]

copy[0][0] = 999
print(matrix[0][0])  # 999 - OOPS! Changed!

# RIGHT - Use deepcopy
copy = deepcopy(matrix)
copy[0][0] = 999
print(matrix[0][0])  # 1 - Safe!
```

### Mistake 2: Not importing deepcopy

```python
# ERROR - deepcopy not available
copy = deepcopy(matrix)  # NameError: name 'deepcopy' is not defined

# CORRECT - Import first
from copy import deepcopy
copy = deepcopy(matrix)  # Works!
```

### Mistake 3: Confusing when you need it

```python
# Simple list - shallow is fine
lst = [1, 2, 3]
copy = lst.copy()  # OK

# Nested list - MUST use deep
lst = [[1, 2], [3, 4]]
copy = deepcopy(lst)  # Required!
```

## Performance Note

Deep copy is slower than shallow copy because it recursively copies all nested objects:

```python
import time
from copy import deepcopy

large_nested = [[[i for i in range(100)] for _ in range(100)] for _ in range(100)]

# Shallow copy (fast)
start = time.time()
shallow = large_nested[:]
print(f"Shallow: {time.time() - start:.6f} seconds")

# Deep copy (slower, but necessary for nested data)
start = time.time()
deep = deepcopy(large_nested)
print(f"Deep: {time.time() - start:.6f} seconds")

# Deep copy is slower, but guarantees independence
```

**Performance Trade-off:**
- Shallow copy: Fast, but risky with nested data
- Deep copy: Slower, but completely safe

|----------|-----|------|-----|

| Nested list | Deep | `deepcopy(lst)` | Need independence |

| Complex objects | Deep | `deepcopy(obj)` | Full independence |
