## Important Gotchas

### Gotcha 1: Assignment vs Copy
```python
list1 = [1, 2, 3]
list2 = list1        # Points to same list (alias)
list2.append(4)
print(list1)         # [1, 2, 3, 4] - CHANGED!

# Correct way to copy:
list2 = list1.copy()        # Shallow copy
list2 = list1[:]            # Shallow copy
from copy import deepcopy
list2 = deepcopy(list1)     # Deep copy (for nested lists)
```

### Gotcha 2: Default Mutable Arguments
```python
def bad_function(items=[]):  # DANGER!
    items.append(1)
    return items

print(bad_function())  # [1]
print(bad_function())  # [1, 1] - Same list reused!

# Correct:
def good_function(items=None):
    if items is None:
        items = []
    items.append(1)
    return items
```

### Gotcha 3: List Comprehension Scope
```python
# x is accessible outside (unlike some languages)
[x for x in range(5)]
print(x)  # 4 - x still exists!
```

## Best Practices

1. Use `append()` for adding single items
2. Use `extend()` for adding multiple items (not `+` in loops)
3. Use list comprehension for creating new lists
4. Use `in` operator to check existence (not `index()` with try/except)
5. Use `enumerate()` when you need both index and value
6. Use slicing `[:]` instead of `list()` for shallow copy
7. Be careful with mutable default arguments
8. Don't modify a list while iterating over it
9. Use `sorted()` when you need a new list; `sort()` for in-place
10. Remember: lists are mutable, strings are not

## String Immutability - THE BIG GOTCHA!

**Once you create a string, you CANNOT change it:**

```python
text = "hello"

# THIS DOES NOT WORK:
# text[0] = 'H'  # TypeError: 'str' object does not support item assignment

# Instead, you must create a new string:
text = "H" + text[1:]  # "Hello"

# OR use methods that return new strings:
text = text.replace("h", "H")  # "Hello"
text = text.upper()  # "HELLO"
```

**Every operation returns a NEW string - the original is unchanged:**
```python
text = "hello"
text.upper()     # Returns "HELLO" but does NOT modify text
print(text)      # Still "hello" - unchanged!

# You must assign the result:
text = text.upper()  # Now text is "HELLO"
```

## String vs List vs Tuple

| Property | String | List | Tuple |
|----------|--------|------|-------|
| Mutable | NO | YES | NO |
| Create | `"hello"` | `[1, 2]` | `(1, 2)` |
| Index | `str[0]` | `lst[0]` | `tpl[0]` |
| Slice | `str[1:3]` | `lst[1:3]` | `tpl[1:3]` |
| Modify | ERROR | `lst[0] = 5` | ERROR |
| Add element | `str + "x"` | `lst.append()` | `tpl + (1,)` |
| Performance | Faster | Slower | Fastest |
| Memory | Less | More | Less |

## Time Complexity of String Operations

| Operation | Complexity | Notes |
|-----------|-----------|-------|
| Access character | O(1) | Direct access by index |
| Slicing | O(k) | k is length of slice |
| Concatenation | O(n+m) | Creates new string |
| Repetition | O(n*k) | k is repeat count |
| find/index | O(n*m) | Worst case; n is string, m is substring |
| replace | O(n) | Single pass through string |
| split | O(n) | Single pass through string |
| join | O(n) | n is total characters |
| upper/lower | O(n) | Processes each character |
| startswith/endswith | O(k) | k is prefix/suffix length |
| in (substring) | O(n*m) | Worst case |

## Important Gotchas

### Gotcha 1: String is Immutable
```python
text = "hello"
text[0] = "H"  # TypeError - cannot modify!

# You must create a new string:
text = "H" + text[1:]  # Works
# OR
text = text.replace("h", "H")  # Works
```

### Gotcha 2: String + Integer Error
```python
text = "The answer is "
number = 42

# This throws TypeError:
# result = text + number

# You must convert to string:
result = text + str(number)  # Works
```

### Gotcha 3: Escape Characters
```python
# Quotes inside strings
text1 = 'She said "hello"'  # Works
text2 = "She said \"hello\""  # Also works (escaped)
text3 = 'It\'s'  # "It's" (escaped single quote)

# Common escape sequences
"\n"   # Newline
"\t"   # Tab
"\\"   # Backslash
"\""   # Double quote
"'"    # Single quote
```

### Gotcha 4: String Methods Don't Modify Original
```python
text = "hello"
text.upper()    # Returns "HELLO" but...
print(text)     # Still "hello" - not modified!

# Must assign result:
text = text.upper()  # Now text is "HELLO"
```

### Gotcha 5: split() vs split(" ")
```python
text = "hello  world  python"

text.split()       # ['hello', 'world', 'python'] (splits by any whitespace)
text.split(" ")    # ['hello', '', 'world', '', 'python'] (splits by literal space)
```

### Gotcha 6: Comparing Strings
```python
"10" > "9"    # False! (string comparison, not numeric)
int("10") > int("9")  # True (numeric comparison)

"abc" < "xyz"  # True (alphabetical order)
```

## Best Practices

1. Use f-strings for formatting (most readable)
2. Use `strip()` when reading user input or from files
3. Use `split()` without arguments for flexible word splitting
4. Use `join()` for concatenating multiple strings (not `+`)
5. Use `in` operator to check if substring exists
6. Use `startswith()` and `endswith()` instead of slicing for comparisons
7. Always convert to string before concatenating with `+`
8. Remember strings are immutable - operations return new strings
9. Use `lower()` or `upper()` for case-insensitive comparisons
10. Use `replace()` for simple substitutions, `re` module for complex patterns
